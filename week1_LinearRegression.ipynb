week 1 Linear Regression from Scratch

#Theory

-Linear Regression ek supervised ML algorithm hai.
-Iska Kaam ek **best fit line** draw karna hota hai jisme data points ke beech relation dikh sake.
-Formula: y=mX+c
  - m = slope
  - c = intercept
  - we use **least squares method** use karke slope aur intercept calculate karenge.
  - at end **R^2 Score** nikal kar model ki accuracy check karenge.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 5, 4, 5], dtype=float)
```

```python
X_mean = np.mean(X)
y_mean = np.mean(y)

m = np.sum((X - X_mean) * (y - y_mean)) / np.sum((X - X_mean)**2)
c = y_mean - m * X_mean

print("Slope:", m)
print("Intercept:", c)
```

```python
y_pred = m * X + c

plt.scatter(X, y, color="red", label="Actual Data")
plt.plot(X, y_pred, color="blue", label="Regression Line")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.legend()
plt.show()
```

```python
ss_total = np.sum((y - y_mean)**2)
ss_res = np.sum((y - y_pred)**2)
r2 = 1 - (ss_res / ss_total)
print("R² Score:", r2)
```

## ✅ Conclusion
- Linear Regression model banaya from scratch (without sklearn).
- Dataset: Study Hours vs Marks.
- Slope (m) aur Intercept (c) calculate kiya.
- Graph plotted with regression line.
- R² Score ≈ 0.6 → matlab model 60% data ko explain kar raha hai.
