```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB

# Sample dataset
data = {
    "email": [
        "Congratulations you won a lottery click here",
        "Your account has been hacked click this link",
        "Meeting scheduled tomorrow",
        "Please review the attached document",
        "Urgent update your bank details now",
        "Project discussion at 3 PM"
    ],
    "label": ["phishing", "phishing", "safe", "safe", "phishing", "safe"]
}

df = pd.DataFrame(data)

# Convert text to numbers
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df["email"])

y = df["label"]

# Train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Train model
model = MultinomialNB()
model.fit(X_train, y_train)

# Test email
test_email = ["Click this link to update your bank account"]

test_vector = vectorizer.transform(test_email)

prediction = model.predict(test_vector)

print("Email:", test_email[0])
print("Prediction:", prediction[0])
```
